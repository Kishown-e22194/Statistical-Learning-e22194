 Probability Measure ($P$): For any subset $A \subseteq \Omega$

$$ P(A) = \sum_{\omega \in A} p(\omega) $$

## Q4) Probability of Defective Items

### 1.

Sample Space ($\Omega$) :

$$ \Omega = [0, 1] $$

$\sigma$-algebra ($\mathfrak{F}$):

$$ \mathfrak{F} = \mathfrak{B}([0, 1]) $$

Probability Measure ($P$):
 for any Borel set $A \in \mathfrak{F}$

 $$ P(A) = \int_A f(x) \,dx = \int_A \frac{1}{B(2, 18)} x(1-x)^{17} \,dx $$

### 2.
Here,

$$X: \Omega \to \mathbb{R}$$

Because of sample space  already consists of the real number outcomes,  

$X$ is  the  function that mapping each outcome to itself

$$ X(\omega) = \omega \quad \text{for all } \omega \in \Omega \quad \text{for all } \omega \in [0, 1] $$


### 3.
$$
f(x) = \begin{cases}
\frac{1}{B(2, 18)} x (1 - x)^{17}, & 0 < x < 1 \\
0, & \text{otherwise}
\end{cases}
$$

### 4.

To be a valid PDF

1. Non-negativity:

For $x \in (0, 1)$  

$$x > 0$$ & $$(1-x)^{17} > 0$$

 The Beta function $B(2, 18)$ is a positive constant

 Thus, $f(x) \ge 0$ everywhere


2.  Integrates to 1:

$$B(\alpha, \beta) = \int_0^1 x^{\alpha-1}(1-x)^{\beta-1} \,dx$$

$$ \int_{-\infty}^{\infty} f(x) \,dx = \int_0^1 \frac{1}{B(2, 18)} x^{2-1} (1-x)^{18-1} \,dx $$

$$ = \frac{1}{B(2, 18)} \int_0^1 x^{2-1} (1-x)^{18-1} \,dx = \frac{1}{B(2, 18)} \times B(2, 18) = \mathbf{1} $$

### 5.

For a Beta distribution  $$E[X] = \frac{\alpha}{\alpha + \beta}$$

$$ E[X] = \frac{2}{2 + 18} = \frac{2}{20} = \mathbf 0.10 $$

### 6.

For a Beta distribution,  

$$\mathrm{Var}(X) = \frac{\alpha\beta}{(\alpha + \beta)^2 (\alpha + \beta + 1)}$$

$$ \mathrm{Var}(X) = \frac{(2)(18)}{(2 + 18)^2 (2 + 18 + 1)} $$

$$ \mathrm{Var}(X) = \frac{36}{(20)^2 (21)}  \approx \mathbf{0.00428} $$




### 7.
$$ P(X \le 0.10) = \int_{0}^{0.10} \frac{1}{B(2, 18)} x (1 - x)^{17} \,dx $$


### 8.

$$ P(0.05 \le X \le 0.15) = \int_{0.05}^{0.15} \frac{1}{B(2, 18)} x (1 - x)^{17} \,dx $$


### 9.

The Beta distribution is specifically defined on the continuous interval $[0, 1]$. Because $X$ represents a proportion or percentage of defective items, it is physically bounded between 0 and 1 .

Furthermore, the parameters $\alpha$ and $\beta$ can be tuned to skew the distribution. In this case, $\beta \gg \alpha$ skews the distribution heavily toward the left (closer to 0), which perfectly models a functional factory where defect rates are usually quite low but occasionally spike.


### 10.

Expectation:

Over a long period of time, evaluating many large batches, the manufacturer can expect **10%** of all items produced to be defective.


Variance:

 It means the defect rate does not wildly swing from 2% one day to 80% the next. Almost all batches will have a defect proportion very tightly clustered right around that 10% average.

## Q5) Expected Project Completion Time

### 1.

### 2.
Given $a=4$, $b=10$, and $c=6$

Peak height:
$$\frac{2}{b-a} = \frac{2}{10-4} = \frac{2}{6} = \frac{1}{3}$$

Left slope ($4 \le x \le 6$):

$$\frac{2(x-a)}{(b-a)(c-a)} = \frac{2(x-4)}{(6)(2)} = \frac{x-4}{6}$$

Right slope ($6 < x \le 10$):

$$\frac{2(b-x)}{(b-a)(b-c)} = \frac{2(10-x)}{(6)(4)} = \frac{10-x}{12}$$

Thus, the PDF is
$$
f_X(x) = \begin{cases}
\frac{x-4}{6}, & 4 \le x \le 6 \\
\frac{10-x}{12}, & 6 < x \le 10 \\
0, & \text{otherwise}
\end{cases}
$$




### 3.

$$ \int_4^{10} f_X(x) \,dx = \int_4^6 \frac{x-4}{6} \,dx + \int_6^{10} \frac{10-x}{12} \,dx $$

$$ \int_4^{10} f_X(x) \,dx = \frac{1}{6} \left[ \frac{(x-4)^2}{2} \right]_4^6 + \frac{1}{12} \left[ -\frac{(10-x)^2}{2} \right]_6^{10}$$

$$ \int_4^{10} f_X(x) \,dx = \frac{1}{3} + \frac{2}{3} $$


$$ \int_4^{10} f_X(x) \,dx = 1 $$

### 4.

$$ E[X] = \frac{a + b + c}{3} $$

$$ E[X] = \frac{4 + 10 + 6}{3} \approx 6.67 \text{ hours} $$

$$ \mathrm{Var}(X) = \frac{a^2 + b^2 + c^2 - ab - ac - bc}{18} $$

$$ \mathrm{Var}(X) = \frac{4^2 + 10^2 + 6^2 - (4)(10) - (4)(6) - (10)(6)}{18} $$

$$ \mathrm{Var}(X) \approx 1.56 $$

## Q6) Expected Project Completion Time Using PERT

### 1.


Given that  $a = 8$,  $b = 20$, $c = 11$

since,  $$b - a = 20 - 8 = 12$$

**For $\alpha$:**
$$ \alpha = 1 + 4 \left( \frac{c - a}{b - a} \right) = 1 + 4 \left( \frac{11 - 8}{12} \right) $$

$$ \alpha  = 2 $$

**For $\beta$:**
$$ \beta = 1 + 4 \left( \frac{b - c}{b - a} \right) = 1 + 4 \left( \frac{20 - 11}{12} \right) $$

$$ \beta = 4 $$

### 2.

The support of $X$ is the interval of all possible completion times

$$ \mathbf{x \in [8, 20]} $$

### 3.


$$ f_X(x) = \frac{(x - a)^{\alpha - 1}(b - x)^{\beta - 1}}{B(\alpha, \beta)(b - a)^{\alpha + \beta - 1}} $$

$$ f_X(x) = \frac{(x - 8)^{2 - 1}(20 - x)^{4 - 1}}{B(2, 4)(20 - 8)^{2 + 4 - 1}} $$

$$ f_X(x) = \begin{cases}
\frac{(x - 8)(20 - x)^3}{B(2, 4) \cdot 12^5}, & 8 \le x \le 20 \\
0, & \text{otherwise}
\end{cases} $$

### 4.

* Sample Space ($\Omega$): It is possible project completion days

$$ \Omega = [8, 20] $$
* $\sigma$-algebra ($\mathfrak{F}$): The Borel $\sigma$-algebra on this interval

$$ \mathfrak{F} = \mathfrak{B}([8, 20]) $$

* Probability Measure ($P$): For any set $A \in \mathfrak{F}$

$$ P(A) = \int_A f_X(x) \,dx $$

### 5.
 Since the sample space already consists of the numerical days

$X$ is the identity function:
$$ X(\omega) = \omega \quad \text{for all } \omega \in [8, 20] $$

### 6.

$$ E[X] = \frac{a + 4c + b}{6} $$

$$ E[X]  = \frac{8 + 44 + 20}{6} $$

$$ E[X] = 12 \text{ days} $$

### 7.

$$ \mathrm{Var}(X) = \frac{(E[X] - a)(b - E[X])}{7} $$

$$ \mathrm{Var}(X) = \frac{(12 - 8)(20 - 12)}{7} $$

$$ \mathrm{Var}(X)   \approx \ 4.57 $$

### 8.

$$ \sigma = \sqrt{\frac{32}{7}} \approx \ 2.14 \text{ days} $$

### 9.

The Triangular distribution forms a sharp peak at the most likely value and connects straight to the extremes. This heavily overemphasizes the probability of the extreme worst case or best case scenarios occurring.

The PERT distribution uses a smooth curve that concentrates more probability mass around the expected value  and drops off smoothly toward the tails. In software integration, complete catastrophes  or perfect runs  are highly unlikely a smooth drop-off models human estimation and real world project delays much more realistically than rigid straight lines.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import beta, triang
import seaborn as sns

# Setup theme
plt.style.use('dark_background')
sns.set_context("talk", font_scale=0.9)

# Project Parameters
a = 8
c = 11
b = 20
scale = b - a

# Create x-axis values from 7 to 21 for a clean plot
x = np.linspace(7, 21, 500)


# 1. PERT Distribution
alpha_pert = 2
beta_pert = 4
pdf_pert = beta.pdf(x, alpha_pert, beta_pert, loc=a, scale=scale)


# 2. Triangular Distribution
c_shape = (c - a) / scale
pdf_triang = triang.pdf(x, c_shape, loc=a, scale=scale)


# Plotting Architecture
fig, ax = plt.subplots(figsize=(12, 6), facecolor='#121212')
ax.set_facecolor('#1a1a1a')

# Plot Triangular
ax.plot(x, pdf_triang, color='#00f0ff', lw=3, linestyle='--', label='Triangular Distribution')
ax.fill_between(x, pdf_triang, alpha=0.1, color='#00f0ff')

# Plot PERT
ax.plot(x, pdf_pert, color='#ff0055', lw=3, label='PERT Distribution (Smoothed)')
ax.fill_between(x, pdf_pert, alpha=0.2, color='#ff0055')

# Add key markers
ax.axvline(c, color='#aaaaaa', linestyle=':', lw=2, label=f'Mode (Most Likely) = {c}')
ax.axvline(12, color='#ffaa00', linestyle='-.', lw=2, label=f'PERT Expected Value = 12')

# Formatting
ax.set_title("PERT vs. Triangular Distribution in Project Estimation", color='white', pad=15, fontweight='bold')
ax.set_xlabel("Days to Complete Task (x)", color='#aaaaaa')
ax.set_ylabel("Probability Density f(x)", color='#aaaaaa')
ax.set_xlim(7, 21)
ax.set_ylim(0, 0.2)

# Grid and Spines
ax.grid(color='#333333', linestyle='--', linewidth=0.5)
for spine in ax.spines.values():
    spine.set_color('#444444')
ax.tick_params(colors='#aaaaaa')

# Legend
leg = ax.legend(facecolor='#1a1a1a', edgecolor='#444444', loc='upper right')
for text in leg.get_texts(): text.set_color('white')

plt.tight_layout()
plt.show()
```
