### **3.**

By the Doob-Dynkin lemma, the conditional expectation must be a function of  observation.

$$E[X \mid Y] = g(Y)$$

Because $\sigma(Y)$ limits  system to only distinguishing between two states  estimator $g(Y)$ is mathematically forbidden from outputting different values for $\omega_1$ and $\omega_2$. It must output a single constant estimate  when $Y=0$ and another constant estimate  when $Y=1$

$$
E[X\mid Y](\omega)=
\begin{cases}
a, & \omega\in\{\omega_1,\omega_2\},\\
b, & \omega\in\{\omega_3,\omega_4\},
\end{cases}
$$


### **4.**

To find $a$ and $b$,

On the set $\{\omega_1,\omega_2\}$,

$$
\int_{\{\omega_1,\omega_2\}} E[X\mid Y]\,dP=\int_{\{\omega_1,\omega_2\}} X\,dP.
$$

$$
a(0.50+0.10)=0\cdot 0.50 + 1\cdot 0.10
$$

$$
0.60a=0.10,
$$

hence

$$
a=\frac{0.10}{0.60}=\frac{1}{6}
$$

On the set $\{\omega_3,\omega_4\}$,

$$
\int_{\{\omega_3,\omega_4\}} E[X\mid Y]\,dP=\int_{\{\omega_3,\omega_4\}} X\,dP.

$$
b(0.20+0.20)=0\cdot 0.20 + 1\cdot 0.20
$$

$$
0.40b=0.20
$$

hence

$$
b=\frac{0.20}{0.40}=\frac{1}{2}
$$

Therefore

$$
E[X\mid Y](\omega)=
\begin{cases}
\dfrac{1}{6}, & \omega\in\{\omega_1,\omega_2\},\\
\dfrac{1}{2}, & \omega\in\{\omega_3,\omega_4\}.
\end{cases}
$$

Equivalently, as a function of $Y$,

$$
E[X\mid Y]=
\frac{1}{6} \mathbb{1}_{\{Y=0\}} + \frac{1}{2} \mathbb{1}_{\{Y=1\}}=
\begin{cases}
\dfrac{1}{6}, & Y=0,\\
\dfrac{1}{2}, & Y=1.
\end{cases}
$$

# **Q2) Proof of the Conditional Density of the Multivariate Gaussian**


$$f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x} \mid \mathbf{y}) = \frac{f_{\mathbf{X}, \mathbf{Y}}(\mathbf{x}, \mathbf{y})}{f_{\mathbf{Y}}(\mathbf{y})}$$



Let the precision matrix $\mathbf{\Lambda}$ be the inverse of the covariance matrix $\boldsymbol{\Sigma}$.

$$\boldsymbol{\Sigma}^{-1} = \mathbf{\Lambda} = \begin{bmatrix} \mathbf{\Lambda}_{XX} & \mathbf{\Lambda}_{XY} \\ \mathbf{\Lambda}_{YX} & \mathbf{\Lambda}_{YY} \end{bmatrix}$$

Using standard block matrix inversion identities,

$$\mathbf{\Lambda}_{XX} = (\boldsymbol{\Sigma}_{XX} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX})^{-1}$$

$$\mathbf{\Lambda}_{XX} = \boldsymbol{\Sigma}_{X|Y}^{-1}$$


$$\mathbf{\Lambda}_{XY} = -\mathbf{\Lambda}_{XX}\boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1} = -\boldsymbol{\Sigma}_{X|Y}^{-1}\boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}$$



$$Q = \begin{bmatrix} \mathbf{x} - \boldsymbol{\mu}_X \\ \mathbf{y} - \boldsymbol{\mu}_Y \end{bmatrix}^T \begin{bmatrix} \mathbf{\Lambda}_{XX} & \mathbf{\Lambda}_{XY} \\ \mathbf{\Lambda}_{YX} & \mathbf{\Lambda}_{YY} \end{bmatrix} \begin{bmatrix} \mathbf{x} - \boldsymbol{\mu}_X \\ \mathbf{y} - \boldsymbol{\mu}_Y \end{bmatrix}$$

$$Q = (\mathbf{x} - \boldsymbol{\mu}_X)^T \mathbf{\Lambda}_{XX} (\mathbf{x} - \boldsymbol{\mu}_X) + 2(\mathbf{x} - \boldsymbol{\mu}_X)^T \mathbf{\Lambda}_{XY} (\mathbf{y} - \boldsymbol{\mu}_Y) + (\text{terms with only } \mathbf{y})$$




$$Q_x = (\mathbf{x} - \boldsymbol{\mu}_X)^T \mathbf{\Lambda}_{XX} (\mathbf{x} - \boldsymbol{\mu}_X) - 2(\mathbf{x} - \boldsymbol{\mu}_X)^T \mathbf{\Lambda}_{XX} \left[ \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}(\mathbf{y} - \boldsymbol{\mu}_Y) \right]$$

let's define the shift caused by $\mathbf{y}$ as a new vector $\mathbf{m}$:
$$\mathbf{m} = \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}(\mathbf{y} - \boldsymbol{\mu}_Y)$$


$$Q_x = (\mathbf{x} - \boldsymbol{\mu}_X)^T \boldsymbol{\Sigma}_{X|Y}^{-1} (\mathbf{x} - \boldsymbol{\mu}_X) - 2(\mathbf{x} - \boldsymbol{\mu}_X)^T \boldsymbol{\Sigma}_{X|Y}^{-1} \mathbf{m}$$


$$Q_x = (\mathbf{x} - \boldsymbol{\mu}_X - \mathbf{m})^T \boldsymbol{\Sigma}_{X|Y}^{-1} (\mathbf{x} - \boldsymbol{\mu}_X - \mathbf{m}) - \mathbf{m}^T \boldsymbol{\Sigma}_{X|Y}^{-1} \mathbf{m}$$

Let's define our new shifted mean, $\boldsymbol{\mu}^*$:
$$\boldsymbol{\mu}^* = \boldsymbol{\mu}_X + \mathbf{m} = \boldsymbol{\mu}_X + \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}(\mathbf{y} - \boldsymbol{\mu}_Y)$$


$$Q_x = (\mathbf{x} - \boldsymbol{\mu}^*)^T \boldsymbol{\Sigma}_{X|Y}^{-1} (\mathbf{x} - \boldsymbol{\mu}^*) + (\text{leftover terms depending only on } \mathbf{y})$$

$$|\boldsymbol{\Sigma}| = |\boldsymbol{\Sigma}_{YY}| \cdot |\boldsymbol{\Sigma}_{XX} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}| = |\boldsymbol{\Sigma}_{YY}| \cdot |\boldsymbol{\Sigma}_{X|Y}|$$


$$f_{\mathbf{X}, \mathbf{Y}}(\mathbf{x}, \mathbf{y}) = \left[ \frac{1}{\sqrt{(2\pi)^n |\boldsymbol{\Sigma}_{X|Y}|}} \exp\left(-\frac{1}{2}(\mathbf{x} - \boldsymbol{\mu}^*)^T \boldsymbol{\Sigma}_{X|Y}^{-1} (\mathbf{x} - \boldsymbol{\mu}^*)\right) \right] \times \left[ \text{Function of } \mathbf{y} \right]$$

By the definition of conditional probability, the bracketed term containing $\mathbf{x}$ **must** be the conditional density $f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x} \mid \mathbf{y})$, and the leftover $\mathbf{y}$ terms perfectly form the marginal density $f_{\mathbf{Y}}(\mathbf{y})$.

This concludes the proof that:
$$f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x} \mid \mathbf{y}) = \frac{1}{\sqrt{(2\pi)^n |\boldsymbol{\Sigma}_{X|Y}|}} \exp\left(-\frac{1}{2}(\mathbf{x} - \boldsymbol{\mu}^*)^T \boldsymbol{\Sigma}_{X|Y}^{-1} (\mathbf{x} - \boldsymbol{\mu}^*)\right)$$
*(Where $|\mathbf{S}_{XX}|$ in the assignment represents $|\boldsymbol{\Sigma}_{X|Y}|$)*.

# Q3) Continuous Fault Monitoring of a Spindle Motor: Solution

### **1.**
Since the internal temperature $X$ and the mechanical load $Q$ are independent normal random variables, their sum $Y = X + Q$ is also normally distributed.

Mean of $Y$: $$E[Y] = E[X] + E[Q] = 0 + 0 = \mathbf{0}$$

Variance of $Y$:

$$\mathrm{Var}(Y) = \mathrm{Var}(X) + \mathrm{Var}(Q) = 1 + 4 = \mathbf{5}$$

Covariance of $(X, Y)$: $$\mathrm{Cov}(X, Y) = E[XY] - E[X]E[Y]$$

Since $E[X]=0$ and $E[Y]=0$

$$E[X(X+Q)] = E[X^2 + XQ] = E[X^2] + E[X]E[Q]$$

Since $E[X^2] = \mathrm{Var}(X) = 1$ and $E[X]E[Q] = 0$

$$\mathrm{Cov}(X, Y) = \mathbf{1}$$

The Joint Distribution is:
$$ \begin{pmatrix} X \\ Y \end{pmatrix} \sim \mathscr{N} \left( \begin{pmatrix} 0 \\ 0 \end{pmatrix}, \begin{pmatrix} 1 & 1 \\ 1 & 5 \end{pmatrix} \right) $$


### **2.**
$X$ is **not** $\sigma(Y)$-measurable because there is no deterministic function $g$ where $X = g(Y)$. For any single observed power reading ($Y=y$), there are infinitely many combinations of temperature ($x$) and torque ($q$) that could equal $y$.



### **3.**

$$E[X \mid Y = y] = \mu_X + \frac{\mathrm{Cov}(X,Y)}{\mathrm{Var}(Y)}(y - \mu_Y)$$
$$E[X \mid Y] = 0 + \frac{1}{5}(Y - 0)$$
**$$E[X \mid Y] = 0.2Y$$**

Signal-to-Total-Variance Interpretation:

The coefficient $0.2$ is the ratio of the signal variance ($\mathrm{Var}(X)=1$) to the total observed variance ($\mathrm{Var}(Y)=5$). We only "trust" 20% of the power fluctuation as being driven by temperature.



### **4.**
Let estimate be $\widehat{X} = 0.2Y$

The residual error is $Z = X - 0.2Y$
Verify Mean is Zero:

$$E[Z] = E[X] - 0.2E[Y] = 0 - 0 = \mathbf{0}$$


Verify Orthogonality (Uncorrelated with $Y$):

  $$\mathrm{Cov}(Z, Y) = E[ZY] - E[Z]E[Y] = E[ZY] - 0$$

  $$E[(X - 0.2Y)Y] = E[XY] - 0.2E[Y^2]$$

  We already found $E[XY] = \mathrm{Cov}(X,Y) = 1$, and $E[Y^2] = \mathrm{Var}(Y) = 5$

  $$\mathrm{Cov}(Z, Y) = 1 - 0.2(5) = 1 - 1 = \mathbf{0}$$
This proves we have extracted 100% of the useful temperature information from the power sensor.


### **5.**

Recall $h(\mathscr{N}(\mu, \sigma^2)) = \frac{1}{2}\ln(2\pi e \sigma^2)$

Marginal $h(X)$: With $\sigma_X^2 = 1$:

$$h(X) = \frac{1}{2}\ln(2\pi e) \approx 1.4189 \text{ nats}$$

Conditional $h(X|Y)$:The conditional variance is
$$\sigma_{X|Y}^2 = \sigma_X^2 - \frac{\text{Cov}(X,Y)^2}{\text{Var}(Y)} = 1 - \frac{1}{5} = 0.8$$

$$h(X|Y) = \frac{1}{2}\ln(2\pi e \cdot 0.8) \approx 1.3073 \text{ nats}$$


### **6.**
Mutual Information is the reduction in uncertainty:

$$I(X; Y) = h(X) - h(X \mid Y)$$

$$I(X; Y) = \frac{1}{2}\ln(2\pi e) - \frac{1}{2}\ln(2\pi e \cdot 0.8) = \frac{1}{2}\ln\left(\frac{1}{0.8}\right) = \frac{1}{2}\ln(1.25)$$

$$I(X; Y) \approx 0.1116 nats$$

Percentage Removed:

The variance was reduced from $1.0$ to $0.8$, meaning observing the power meter removes exactly 20% of the uncertainty about the temperature.

If Mechanical Noise was $100$:

The variance of $Y$ would be $101$. The conditional variance would be $1 - (1/101) \approx 0.99$. The Mutual Information would drop to near zero, as the tiny temperature signal would be completely "drowned out" by the massive mechanical noise.



### **7.**

If the manager simply uses the raw reading ($X \approx Y$), the error variance is massive:

$$E[(X-Y)^2] = E[(-Q)^2] = \mathrm{Var}(Q) = \mathbf{4}$$

By using the conditional expectation ($E[X \mid Y] = 0.2Y$), the error variance is explicitly minimized to $0.8$.

The concept of "Shrinkage":

The factor $0.2$ is called a shrinkage estimator. It tells us that when a sensor is noisy, the "best" guess is to be conservative and pull the reading back toward the historical average (zero). If the power spikes, we don't immediately assume the tool is melting; we assume most of that spike is just the material being tough.
