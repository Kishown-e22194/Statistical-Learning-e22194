# 🧼 DataWash Inspector: End-to-End Cleaning & EDA Engine

DataWash Inspector replaces hundreds of lines of tedious Pandas and Matplotlib code with a single, chainable, automated pipeline. It systematically cleans datasets and generates a stunning, serverless Exploratory Data Analysis (EDA) dashboard directly inside Google Colab.

### Part 1: Automated Data Cleaning Pipeline
1. **Smart Ingestion:** Automatically detects the Colab environment to trigger the native `google.colab.files.upload()` UI. Supports `.csv`, `.xlsx`, `.json`, `.parquet`, and `.html`.
2. **Garbage Sanitization:** Hunts down inconsistent nulls (`?`, `N/A`, `#DIV/0!`), strips invisible whitespaces, and deduplicates column headers.
3. **Type Inference:** Safely coerces numeric strings into mathematical floats, and automatically parses strict datetimes to unlock time-series capabilities.
4. **Smart Imputation:** Uses statistical strategies (`strategy='auto'`) to fill missing values using the mean (for numbers) and the mode (for categories).
5. **Text Standardization:** Forcefully standardizes text casing and spacing to ensure categorical aggregations are perfectly uniform.
6. **Outlier Handling:** Uses the IQR (Interquartile Range) method to mathematically identify and flag/drop statistical outliers.
7. **Memory Optimization:** Dynamically downcasts numeric columns (`float64` → `float32`, `int64` → `int8/16/32`), reducing RAM usage by up to 70% to prevent Colab crashes.

### Part 2: Automated EDA & Plotting Engine
Once the data is pristine, the built-in `PlotFactory` engine takes over, replacing manual plotting with a Plotly-powered, highly interactive UI:
1. **Premium Glassmorphic Aesthetics:** Charts feature a premium dark mode (`#1e1e2f`), transparent gridlines, and sit inside "glassmorphic" CSS cards with blurred backgrounds. Every chart is fully interactive (hover, zoom, pan, export).
2. **Univariate Analysis:** Automatically generates a dual-view subplot for every numeric column: a probability distribution Histogram stacked perfectly above a Box Plot to instantly reveal quartile bounds.
3. **Categorical Frequency:** Generates auto-sorted bar charts for text columns with 45-degree smart labels to prevent text overlap.
4. **Universal Association Heatmap:** Standard correlation fails on text. DataWash solves this by computing a universal matrix using *Pearson* (Num/Num), *Cramer's V* (Cat/Cat), and the *Correlation Ratio* (Num/Cat), plotted as one massive interactive heatmap.
5. **Smart Relationship Routing:** The engine algorithmically isolates the Top 5 strongest variable relationships and dynamically picks the best plot:
    * *Num vs Num:* 2D Density Contour / Scatter Plots
    * *Cat vs Num:* Side-by-Side Box Plots
    * *Cat vs Cat:* 100% Stacked Bar Charts
6. **Colab Proxy Architecture:** To prevent browser memory crashes from massive iframe strings, the library intelligently down-samples visual data to 1,500 rows and spins up a background Python `http.server`. It securely injects the dashboard via Google Colab's native proxy API (`serve_kernel_port_as_iframe`), enabling offline HTML exports and lightning-fast rendering.
