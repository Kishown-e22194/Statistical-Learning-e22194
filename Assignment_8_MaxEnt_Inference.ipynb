## Q. Statistical Mechanics of the Quantum Oscillator


## 1. Partition Function

The canonical partition function is

$$
Z(T)=\sum_{j=0}^{\infty}e^{-\beta E_j}
$$

Substitute the energy eigenvalues:

$$
Z(T)=\sum_{j=0}^{\infty}e^{-\beta\hbar\omega\left(j+\frac12\right)}
$$



$$
Z(T)=e^{-\frac12\beta\hbar\omega}
\sum_{j=0}^{\infty}e^{-j\beta\hbar\omega}
$$

This is a geometric series with common ratio

$$
r=e^{-\beta\hbar\omega}
$$

Since $T>0$, $\beta>0$, so

$$
0<e^{-\beta\hbar\omega}<1
$$

Therefore, the series converges:

$$
\sum_{j=0}^{\infty}r^j=\frac{1}{1-r}
$$

Hence,

$$
Z(T)=e^{-\frac12\beta\hbar\omega}
\frac{1}{1-e^{-\beta\hbar\omega}}
$$

Therefore,

$$
Z(T)=
\frac{
e^{-\frac{\hbar\omega}{2k_BT}}
}{
1-e^{-\frac{\hbar\omega}{k_BT}}
}
$$


## 2. Helmholtz Free Energy

The Helmholtz free energy is

$$
F(T)=-k_BT\ln Z
$$

From the partition function,

$$
Z=
\frac{
e^{-\frac12\beta\hbar\omega}
}{
1-e^{-\beta\hbar\omega}
}
$$



$$ln Z=
-\frac12\beta\hbar\omega\ln\left(1-e^{-\beta\hbar\omega}\right)
$$

Therefore,

$$
F(T)=-k_BT
\left[
-\frac12\beta\hbar\omega\ln\left(1-e^{-\beta\hbar\omega}\right)
\right]
$$

Since

$$
\beta=\frac{1}{k_BT}
$$


$$
F(T)=
\frac{\hbar\omega}{2}
+
k_BT\ln\left(1-e^{-\frac{\hbar\omega}{k_BT}}\right)
$$

### Internal Energy

The internal energy is

$$
U(T)=
-\frac{\partial \ln Z}{\partial \beta}
$$



$$
\ln Z=
-\frac12\beta\hbar\omega\ln\left(1-e^{-\beta\hbar\omega}\right)
$$

Let

$$
a=\hbar\omega
$$

Then

$$
\ln Z=
-\frac12\beta a\ln\left(1-e^{-\beta a}\right)
$$

Differentiate with respect to $\beta$:

$$
\frac{\partial \ln Z}{\partial \beta}=-\frac{a}{2}-\frac{ae^{-\beta a}}{1-e^{-\beta a}}
$$

Therefore,

$$
U=
-\frac{\partial \ln Z}{\partial \beta}
$$

So,

$$
U=
\frac{a}{2}
+
\frac{ae^{-\beta a}}{1-e^{-\beta a}}
$$

Since

$$
\frac{e^{-\beta a}}{1-e^{-\beta a}}=\frac{1}{e^{\beta a}-1}
$$

$$
U(T)=
\frac{\hbar\omega}{2}+\frac{\hbar\omega}{e^{\frac{\hbar\omega}-{k_BT}}-1}
$$

###3. Entropy

Entropy is given by

$$
S(T)=
-\left(\frac{\partial F}{\partial T}\right)
$$

We have

$$
F(T)=
\frac{\hbar\omega}{2}
+
k_BT\ln\left(1-e^{-\frac{\hbar\omega}{k_BT}}\right)
$$

Let

$$
x=\frac{\hbar\omega}{k_BT}
$$

Then

$$
F(T)=
\frac{\hbar\omega}{2}
+
k_BT\ln(1-e^{-x})
$$

Differentiating with respect to $T$

$$
\frac{\partial F}{\partial T}=k_B\ln(1-e^{-x})-k_B\frac{x}{e^x-1}
$$

Therefore,

$$
S(T)=-k_B\ln(1-e^{-x})+k_B\frac{x}{e^x-1}
$$

Substitute back

$$
x=\frac{\hbar\omega}{k_BT}
$$

So,

$$
S(T)=k_B\left[\frac{\frac{\hbar\omega}{k_BT}}{e^{\frac{\hbar\omega}{k_BT}}-1}
-\ln\left(1-e^{-\frac{\hbar\omega}{k_BT}}\right)\right]
$$



### Verification of $U=F+TS$



$$
F=
\frac{\hbar\omega}{2}
+
k_BT\ln(1-e^{-x})
$$

and

$$
S=
k_B
\left[
\frac{x}{e^x-1}-\ln(1-e^{-x})\right]
$$

Therefore,

$$
TS=k_BT\left[\frac{x}{e^x-1}-\ln(1-e^{-x})\right]
$$

Now,

$$
F+TS= \frac{\hbar\omega}{2} + k_BT\ln(1-e^{-x}) +k_BT\left[\frac{x}{e^x-1}-\ln(1-e^{-x})\right]
$$

The logarithm terms cancel:

$$
F+TS=\frac{\hbar\omega}{2}+k_BT\frac{x}{e^x-1}
$$

Since

$$
x=\frac{\hbar\omega}{k_BT}
$$

we have

$$
k_BTx=\hbar\omega
$$

Therefore,

$$
F+TS=
\frac{\hbar\omega}{2}
+
\frac{\hbar\omega}{e^x-1}
$$

Hence,

$$
F+TS=U
$$



### 4. Heat Capacity

The constant-volume heat capacity is

$$
C_V=
\left(\frac{\partial U}{\partial T}\right)
$$

The internal energy is

$$
U(T)=
\frac{\hbar\omega}{2}
+
\frac{\hbar\omega}{e^x-1}
$$

where

$$
x=\frac{\hbar\omega}{k_BT}
$$

The zero-point energy term is constant, so it does not contribute to heat capacity.

Using

$$
\frac{dx}{dT}=-\frac{x}{T}
$$

we get

$$
C_V=
k_Bx^2
\frac{e^x}{(e^x-1)^2}
$$

Substitute back

$$
x=\frac{\hbar\omega}{k_BT}
$$

Therefore,

$$
C_V=k_B\left(\frac{\hbar\omega}{k_BT}\right)^2\frac{e^{\frac{\hbar\omega}{k_BT}}
}{\left(e^{\frac{\hbar\omega}{k_BT}}-1\right)^2}
$$



### 5.Hyperbolic Form of Heat Capacity


$$
C_V=k_Bx^2\frac{e^x}{(e^x-1)^2}
$$

Now,

$$
e^x-1=e^{x/2}\left(e^{x/2}-e^{-x/2}\right)
$$

Therefore,

$$
(e^x-1)^2=e^x\left(e^{x/2}-e^{-x/2}\right)^2
$$

So,

$$
\frac{e^x}{(e^x-1)^2} =\frac{1}{\left(e^{x/2}-e^{-x/2}\right)^2
}
$$

Using

$$
\sinh(y)=\frac{e^y-e^{-y}}{2}
$$

we get

$$
e^{x/2}-e^{-x/2}=2\sinh\left(\frac{x}{2}\right)
$$

Hence,

$$
\left(e^{x/2}-e^{-x/2}\right)^2=4\sinh^2\left(\frac{x}{2}\right)
$$

Therefore,

$$
C_V=k_Bx^2\frac{1}{4\sinh^2\left(\frac{x}{2}\right)
}
$$

So,

$$
C_V=k_B\left(\frac{x}{2}\right)^2\frac{1}{\sinh^2\left(\frac{x}{2}\right)}
$$

Substituting

$$
x=\frac{\hbar\omega}{k_BT}
$$

we obtain

$$
C_V=k_B\left(\frac{\hbar\omega}{2k_BT}\right)^2\frac{1}{\sinh^2\left(\frac{\hbar\omega}{2k_BT}\right)}
$$



### High-Temperature Limit

In the high-temperature limit,

$$
k_BT\gg \hbar\omega
$$

So,

$$
x=\frac{\hbar\omega}{k_BT}\ll 1
$$

Using the approximation

$$
\sinh\left(\frac{x}{2}\right)\approx\frac{x}{2}
$$

we get

$$
C_V=k_B\left(\frac{x}{2}\right)^2\frac{1}{\sinh^2(x/2)}
$$

Therefore,

$$
C_V\approx\ k_B\left(\frac{x}{2}\right)^2\frac{1}{(x/2)^2}
$$

So,

$$
C_V\to k_B
$$

This agrees with the classical equipartition theorem.

For a one-dimensional harmonic oscillator, the classical average energy is

$$
U_{\text{classical}}=k_BT
$$

because it has two quadratic energy contributions:

$$
\frac12 k_BT
$$

from kinetic energy and

$$
\frac12 k_BT
$$

from potential energy.

Therefore,

$$
C_V=\frac{\partial U_{\text{classical}}}{\partial T}=k_B
$$

So the quantum harmonic oscillator approaches the classical result at high temperature.



## Q. The Maximum Entropy Distribution under Mean and Variance Constraints


For a continuous random variable  
$$
h(X)=-\int_{-\infty}^{\infty} f(x)\ln f(x)\,dx
$$

$$
\int_{-\infty}^{\infty} f(x)\,dx=1
$$

$$
\int_{-\infty}^{\infty} x f(x)\,dx=\mu
$$

$$
\int_{-\infty}^{\infty} (x-\mu)^2 f(x)\,dx=\sigma^2
$$


Let
$$
y=x-\mu
$$

Then the mean constraint becomes

$$
\mathbb E[Y]=0
$$

and the variance constraint becomes

$$
\mathbb E[Y^2]=\sigma^2
$$

$$
h(f)=-\int_{-\infty}^{\infty} f(y)\ln f(y)\,dy
$$

subject to
$$
\int_{-\infty}^{\infty} f(y)\,dy=1
$$

$$
\int_{-\infty}^{\infty} y f(y)\,dy=0
$$

$$
\int_{-\infty}^{\infty} y^2 f(y)\,dy=\sigma^2
$$

form the Lagrangian
$$
\mathcal L = -\int f(y)\ln f(y)\,dy -\lambda_0\left(\int f(y)\,dy-1\right) -\lambda_1\left(\int y f(y)\,dy-0\right) -\lambda_2\left(\int y^2 f(y)\,dy-\sigma^2\right)
$$

take the functional derivative with respect to $f(y)$.

Since

$$
\frac{d}{df}\left[-f\ln f\right] = -(\ln f+1)
$$

$$
-(\ln f(y)+1)-\lambda_0-\lambda_1y-\lambda_2y^2=0
$$

Therefore,
$$
\ln f(y) = -1-\lambda_0-\lambda_1y-\lambda_2y^2
$$
Taking exponential on both sides,
$$
f(y) = e^{-1-\lambda_0} e^{-\lambda_1y-\lambda_2y^2}
$$
Let
$$
A=e^{-1-\lambda_0}
$$
Then
$$
f(y)=A e^{-\lambda_1y-\lambda_2y^2}
$$

The constraint is
$$
\mathbb E[Y]=0
$$

The maximum entropy distribution should be centered around $y=0$.

Therefore, there should be no linear shift term in the exponent. So,
$$
\lambda_1=0
$$

Hence,
$$
f(y)=A e^{-\lambda_2y^2}
$$


let use normalization:
$$
\int_{-\infty}^{\infty} f(y)\,dy=1
$$

So,
$$
\int_{-\infty}^{\infty} A e^{-\lambda_2y^2}\,dy=1
$$

by using the Gaussian integral,
$$
\int_{-\infty}^{\infty} e^{-\lambda_2y^2}\,dy = \sqrt{\frac{\pi}{\lambda_2}}
$$



$$
A\sqrt{\frac{\pi}{\lambda_2}}=1
$$

Therefore,
$$
A=\sqrt{\frac{\lambda_2}{\pi}}
$$

Now use the variance constraint:
$$
\int_{-\infty}^{\infty} y^2 f(y)\,dy=\sigma^2
$$

For a density of the form
$$
f(y)=A e^{-\lambda_2y^2}
$$

the variance is
$$
\mathbb E[Y^2]=\frac{1}{2\lambda_2}
$$

Therefore,
$$
\frac{1}{2\lambda_2}=\sigma^2
$$

So,
$$
\lambda_2=\frac{1}{2\sigma^2}
$$

Then
$$
A=\sqrt{\frac{1}{2\pi\sigma^2}}
$$

Thus,
$$
A=\frac{1}{\sqrt{2\pi\sigma^2}}
$$


Since
$$
y=x-\mu
$$

$$
f(x) = \frac{1}{\sqrt{2\pi\sigma^2}}\exp\left(-\frac{(x-\mu)^2}{2\sigma^2}\right)
$$

Therefore,
$$
X\sim \mathcal N(\mu,\sigma^2)
$$

and
$$
f(x)=\frac{1}{\sqrt{2\pi\sigma^2}}\exp\left(-\frac{(x-\mu)^2}{2\sigma^2}\right)
$$



## Q. Multiclass Logistic Regression: Automatic Classification of Emails



The feature vector is
$$

$$
where
$$
x_1=\mathbb{I}(\text{Financial context detected})
$$
$$

$$

$$

The observed feature scenarios are:

| Scenario | Feature vector | $P(\mathbf{x})$ |
| :--- | :--- | :--- |
| A: Pure Financial | $(1,0,0)^T$ | 0.40 |

| D: Access Technical | $(0,1,1)^T$ | 0.15 |
| E: Billing Access | $(1,1,0)^T$ | 0.05 |

The given audit constraints are:
$$
\mathbb{E}[x_1\mathbb{I}(Y=1)]=0.05
$$
$$
\mathbb{E}[x_2\mathbb{I}(Y=1)]=0.15
$$
$$
\mathbb{E}[x_3\mathbb{I}(Y=1)]=0.16
$$

This is solved as a binary classification problem:
$$

$$
Y=1:\text{ Technical}

$$
P(Y=1|\mathbf{x}) = \frac{1}{1+\exp(-\mathbf{w}^T\mathbf{x})}
$$
where
$$

$$

where
$$
\sigma(z)=\frac{1}{1+e^{-z}}

For each scenario:

Therefore,

#### **Final Logistic Regression Model**

## Q. Asset Pricing and Portfolio Optimization


### **Answer: Asset Pricing and Portfolio Optimization using Maximum Entropy**

Let
$$
S=S_T
$$
be the one-month future price of Asset X. Since asset prices are positive,
$$
S>0
$$

$$
\mathbb{E}[S]=102
$$
$$
\mathbb{E}[S^2]=10600
$$
$$
\mathbb{E}[\ln S]=4.60
$$

#### **1. Maximum Entropy Objective**
The differential entropy is
$$
h[p]=-\int_0^\infty p(s)\ln p(s)\,ds
$$
The Maximum Entropy problem is:
$$
\max_{p(s)} \left[ -\int_0^\infty p(s)\ln p(s)\,ds \right]
$$
subject to
$$
\int_0^\infty p(s)\,ds=1
$$
$$
\int_0^\infty s p(s)\,ds=102
$$
$$
\int_0^\infty s^2p(s)\,ds=10600
$$
$$
\int_0^\infty \ln(s)p(s)\,ds=4.60
$$

#### **2. Lagrangian**
 Lagrange multipliers for the constraints:
$$
\lambda_0,\lambda_1,\lambda_2,\lambda_3
$$
The Lagrangian functional is
$$
\mathcal{L} = -\int_0^\infty p(s)\ln p(s)\,ds -\lambda_0\left(\int_0^\infty p(s)\,ds-1\right) -\lambda_1\left(\int_0^\infty sp(s)\,ds-102\right) -\lambda_2\left(\int_0^\infty s^2p(s)\,ds-10600\right) -\lambda_3\left(\int_0^\infty \ln(s)p(s)\,ds-4.60\right)
$$

#### **3. Deriving the Density**

$$
\frac{\delta \mathcal{L}}{\delta p(s)} = -(\ln p(s)+1) -\lambda_0 -\lambda_1s -\lambda_2s^2 -\lambda_3\ln s
$$

$$
-(\ln p(s)+1) -\lambda_0 -\lambda_1s -\lambda_2s^2 -\lambda_3\ln s = 0
$$
So,
$$
\ln p(s) = -1-\lambda_0-\lambda_1s-\lambda_2s^2-\lambda_3\ln s
$$

$$
p(s) = e^{-1-\lambda_0} e^{-\lambda_1s-\lambda_2s^2-\lambda_3\ln s}
$$
Since
$$
e^{-\lambda_3\ln s}=s^{-\lambda_3}
$$

$$
p(s) = C s^{-\lambda_3} e^{-\lambda_1s-\lambda_2s^2}
$$
where $C=e^{-1-\lambda_0}$ is the normalizing constant.
Therefore, the Maximum Entropy density has the form
$$
p(s) = \frac{1}{Z} s^{-\lambda_3} \exp\left(-\lambda_1s-\lambda_2s^2\right), \qquad s>0
$$
where
$$
Z = \int_0^\infty s^{-\lambda_3} \exp\left(-\lambda_1s-\lambda_2s^2\right)\,ds
$$

#### **4. Parameter Conditions**

At large $s$, the term
$$
e^{-\lambda_2s^2}
$$
controls the tail. Therefore
$$
\lambda_2>0
$$
Near $s=0$, the term
$$
s^{-\lambda_3}
$$
controls the behavior. To keep the integral finite near zero
$$
-\lambda_3>-1
$$
so
$$
\lambda_3<1
$$
Hence, a valid density requires
$$
\lambda_2>0,\qquad \lambda_3<1
$$

#### **5. Determining the Lagrange Multipliers**

$$
\int_0^\infty s p(s)\,ds=102
$$
$$
\int_0^\infty s^2 p(s)\,ds=10600
$$
$$
\int_0^\infty \ln(s)p(s)\,ds=4.60
$$
Equivalently, using the partition function $Z$,
$$
-\frac{\partial \ln Z}{\partial \lambda_1}=102
$$
$$
-\frac{\partial \ln Z}{\partial \lambda_2}=10600
$$
$$
-\frac{\partial \ln Z}{\partial \lambda_3}=4.60
$$
 In practice, they are solved numerically using methods such as Newton’s method or numerical optimization.

#### **6. Final Maximum Entropy Distribution**
Therefore, the least-biased risk-neutral density of the future asset price is
$$
p(S_T) = \frac{1}{Z} S_T^{-\lambda_3} \exp\left( -\lambda_1S_T-\lambda_2S_T^2 \right), \qquad S_T>0
$$
where
$$
Z = \int_0^\infty S_T^{-\lambda_3} \exp\left( -\lambda_1S_T-\lambda_2S_T^2 \right) \,dS_T
$$
and the parameters satisfy
$$
\mathbb{E}[S_T]=102
$$
$$
\mathbb{E}[S_T^2]=10600
$$
$$
\mathbb{E}[\ln S_T]=4.60
$$

#### **7. Interpretation**
This density is not forced to be Gaussian. A Gaussian distribution is symmetric and can assign probability to negative prices, which is not suitable for asset prices.
The Maximum Entropy density here is defined only for
$$
S_T>0
$$
and it uses the market information directly.
* The constraint $\mathbb{E}[S_T]=102$ matches the risk-neutral forward price.
* The constraint $\mathbb{E}[S_T^2]=10600$ captures the market volatility.
* The constraint $\mathbb{E}[\ln S_T]=4.60$ captures asymmetric downside sentiment. Because $\ln(S_T)$ becomes very small when $S_T$ falls sharply, this constraint allows the model to reflect crash risk and fat-tail behavior.
